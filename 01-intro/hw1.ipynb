```python
cars_asia_df = df2[df2['origin'] == 'Asia'][['vehicle_weight','model_year']].reset_index(drop = True)
```

```python
X = np.array(X_list)
```

```python
print(X)
```

```python
XTX = X.T.dot(X)
```

```python
XTX
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
print(np.linalg.inv(XTX))
```
